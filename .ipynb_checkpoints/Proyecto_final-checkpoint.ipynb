```python
import pandas as pd
import numpy as np
from scipy.stats import chisquare

# Librerías gráficas
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import matplotlib.pyplot as plt
import seaborn as sns

# Preprocesamiento y Modelado
from sklearn.preprocessing import StandardScaler, OneHotEncoder, LabelEncoder,MinMaxScaler
from sklearn.feature_selection import mutual_info_classif
from sklearn.decomposition import PCA
from sklearn.model_selection import train_test_split

# Configuración global
np.random.seed(1989)
pd.set_option("display.float_format", '{:,.2f}'.format)

print("Bibliotecas importadas correctamente.")
```

```python
class analyze:
    # Inicializamos la clase
    def __init__(self, df):
        self.df = df

    # Método para obtener las estadísticas detalladas
    def stats(self, x):
        # Filtramos nulos para cálculos estadísticos
        series = self.df[x].dropna()
        if series.empty:
            return pd.DataFrame()

        minimo = np.round(np.min(series), 4)
        p01 = np.round(np.percentile(series, 1), 4)
        p05 = np.round(np.percentile(series, 5), 4)
        media = np.round(np.mean(series), 4)
        mediana = np.round(np.median(series), 4)
        p95 = np.round(np.percentile(series, 95), 4)
        p99 = np.round(np.percentile(series, 99), 4)
        maximo = np.round(np.max(series), 4)
        varianza = np.round(np.var(series, ddof=1), 4)
        ds = np.round(np.std(series, ddof=1), 4)
        kurt = series.kurtosis()
        skew = series.skew()

        data = {
            'Estadísticas': ["minimo", "P01%", "P05%", "media", "mediana", "P95%", "P99%", "maximo", "varianza", "desv_std", "kurtosis", "skewness"],
            'Valores': [minimo, p01, p05, media, mediana, p95, p99, maximo, varianza, ds, kurt, skew]
        }
        return pd.DataFrame(data)

    # Método para visualizar gráficamente las variables numéricas
    def plot(self, x):
        fig = make_subplots(rows=1, cols=2, subplot_titles=("Boxplot", "Histograma"))

        # Boxplot
        fig.add_trace(go.Box(y=self.df[x], name="Boxplot", marker_color="#008080"), row=1, col=1)

        # Histograma
        fig.add_trace(go.Histogram(x=self.df[x], name="Density", marker_color="#00a77d", nbinsx=50, histnorm='probability density'), row=1, col=2)

        fig.update_layout(showlegend=False, title_text=f"Distribución de: {x}", height=400, width=900)
        fig.show()

    # Método para graficar la matriz de correlación
    def cor_matrix(self):
        corr_matrix = self.df.corr()
        fig = go.Figure(data=go.Heatmap(
            z=corr_matrix.values,
            x=corr_matrix.columns,
            y=corr_matrix.columns,
            colorscale='Viridis',
            zmin=-1, zmax=1
        ))
        fig.update_layout(title='Matriz de Correlación', width=900, height=700)
        fig.show()

def is_na_df(df):
    na_summary = (
        df.isna().sum()
        .reset_index()
        .rename(columns={"index": "columna", 0: "n_missing"})
    )
    na_summary["total"] = len(df)
    na_summary["prop_missing"] = na_summary["n_missing"] / na_summary["total"]
    na_summary["prop_original"] = 1 - na_summary["prop_missing"]

    na_summary = na_summary[na_summary["n_missing"] > 0].reset_index(drop=True)

    total_missing = df.isna().sum().sum()
    total_elements = df.size
    resumen = pd.DataFrame({
        "columna": ["TOTAL"],
        "n_missing": [total_missing],
        "total": [total_elements],
        "prop_missing": [total_missing / total_elements],
        "prop_original": [1 - total_missing / total_elements]
    })

    return pd.concat([na_summary, resumen], ignore_index=True)

# Función auxiliar para WoE e IV
def calculate_woe_iv(df, feature, target):
    lst = []
    # Si la variable es numérica continua, la discretizamos en deciles
    if df[feature].dtype != 'O' and df[feature].nunique() > 10:
        try:
            df['bin'] = pd.qcut(df[feature], 10, duplicates='drop')
            col_name = 'bin'
        except:
            df['bin'] = df[feature] # Fallback si no puede hacer bins
            col_name = 'bin'
    else:
        df['bin'] = df[feature]
        col_name = 'bin'

    for i in df[col_name].unique():
        val = i
        lst.append({
            'Value': str(val),
            'All': df[df[col_name] == val].count()[feature],
            'Good': df[(df[col_name] == val) & (df[target] == 0)].count()[feature],
            'Bad': df[(df[col_name] == val) & (df[target] == 1)].count()[feature]
        })

    dset = pd.DataFrame(lst)
    dset['Distr_Good'] = dset['Good'] / dset['Good'].sum()
    dset['Distr_Bad'] = dset['Bad'] / dset['Bad'].sum()
    dset['WoE'] = np.log(dset['Distr_Good'] / dset['Distr_Bad'])
    dset = dset.replace({'WoE': {np.inf: 0, -np.inf: 0}}) # Tratar divisiones por cero
    dset['IV'] = (dset['Distr_Good'] - dset['Distr_Bad']) * dset['WoE']

    iv = dset['IV'].sum()
    return iv
```

### ¿Cómo afectarán estos outliers a nuestros 3 modelos y qué debemos hacer?

La presencia de estos valores extremos justifica perfectamente por qué elegimos una mezcla de modelos de árboles y modelos matemáticos espaciales:

1.  **Para Random Forest y Gradient Boosting (XGBoost/LightGBM):**

El algoritmo simplemente hará un corte (split) del tipo `si campaign > 10` y aislará esos casos sin que el valor "56" distorsione las predicciones del resto de los clientes.
2.  **Para Máquinas de Soporte Vectorial (SVM):**
    * **Peligro crítico:** SVM es un algoritmo geométrico que calcula distancias entre puntos para trazar su hiperplano de separación. Un solo cliente con 56 llamadas o 98 años puede "jalar" la frontera de decisión y arruinar el modelo.
    * **Solución a implementar:** Antes de entrenar el SVM, debemos aplicar una técnica llamada **Clipping o Winsorización** a variables como `campaign` y `age`. Esto significa "topar" los valores. Por ejemplo, cualquier valor en `campaign` mayor a 10, lo convertimos en 10. Posteriormente, se debe aplicar el `StandardScaler` o `RobustScaler`.



# Winsorización por Percentiles
La mejor práctica para conjuntos de datos con asimetrías tan extremas (como este dataset bancario) es usar la Winsorización por Percentiles (usualmente del 1% al 99%)

```python
import numpy as np

def winsorizar_percentiles(df, columnas, limite_inf=0.01, limite_sup=0.99):
    # Hacemos una copia para no alterar el dataframe original
    df_w = df.copy()

    for var in columnas:
        # Calculamos los percentiles (ej. 1% y 99%)
        lower = df_w[var].quantile(limite_inf)
        upper = df_w[var].quantile(limite_sup)

        # Aplicamos el recorte (Clipping)
        df_w[var] = np.where(df_w[var] < lower, lower, df_w[var])
        df_w[var] = np.where(df_w[var] > upper, upper, df_w[var])

        print(f"Variable '{var}': recorte aplicado entre {lower:.2f} y {upper:.2f}")

    return df_w

# 1. Definimos las variables que sufren de casos extremos
variables_outliers = ['age', 'campaign', 'previous']

# 2. Ejecutamos la función sobre tu dataframe bank_df
# Guardamos el resultado en bank_df_winsorizado para mantener el orden
bank_df_winsorizado = winsorizar_percentiles(bank_df, variables_outliers)
```

La justificación para utilizar la **Winsorización por Percentiles** (específicamente a los percentiles 1% y 99%) en lugar de otros métodos estadísticos como el Rango Intercuartílico (IQR) o la eliminación de filas, se basa en la intersección de tres factores: la matemática de los modelos que elegimos, la distribución extrema de tus datos bancarios y el sentido de negocio.

Aquí tienes la justificación detallada para tu proyecto:

### 1. Protección geométrica para el modelo SVM
Como vimos en tu documento sobre Máquinas de Soporte Vectorial (SVM), este algoritmo traza un **hiperplano de separación óptima** calculando las distancias espaciales entre los puntos de datos (vectores de soporte).
* **El problema:** Si dejas a un cliente con 56 llamadas en la variable `campaign`, ese punto estará tan alejado geométricamente del resto que "jalará" el hiperplano hacia él, deformando la frontera de decisión y haciendo que el modelo clasifique mal a los clientes normales.
* **La solución:** Al recortar (topar) ese valor al percentil 99, acercamos ese punto al límite de la población "normal", permitiendo que el SVM trace una frontera limpia sin perder el registro del cliente.

### 2. El fracaso del método clásico (IQR) por la asimetría hacia el cero
El método tradicional para recortar outliers es usar el Rango Intercuartílico (IQR). Sin embargo, en este dataset bancario, variables como `previous` (contactos antes de esta campaña) tienen un comportamiento radical: **más del 85% de los clientes tienen un valor de 0.**
* **Por qué falla el IQR:** Si el 85% de los datos es 0, tanto el Cuartil 1 (25%) como el Cuartil 3 (75%) equivalen a 0. El cálculo del IQR daría 0, y el límite superior sería 0. Si usaras ese método, el código convertiría todos los contactos previos en 0, **destruyendo por completo la variable y su poder predictivo**.
* **Por qué triunfan los Percentiles:** Al usar el percentil 99, el código ignora la concentración masiva en el cero y busca el valor real donde se encuentra el 1% más alto de la población (por ejemplo, 3 o 4 contactos), aislando únicamente las verdaderas anomalías sin destruir la varianza de la variable.

### 3. Preservación del valor de negocio en los extremos (No eliminar filas)
La reacción intuitiva de muchos analistas es simplemente borrar (hacer *drop*) de las filas con valores atípicos.
* **Justificación del recorte en `age`:** El percentil 99 recorta las edades máximas (que llegan hasta 98 años) y las topa alrededor de los 71 años. Si hubiéramos borrado a los clientes mayores de 70 años, habríamos eliminado a gran parte del grupo de "jubilados" (`job = retired`). Irónicamente, en el sector bancario, los jubilados suelen tener el capital y la disposición para contratar depósitos a plazo. **Winsorizar nos permite conservar la información valiosa de que el cliente aceptó o rechazó el producto, sin que su edad extrema rompa las matemáticas del modelo.**
* **Justificación del recorte en `campaign`:** Topar las llamadas a ~14 (percentil 99) en lugar del máximo de 56 refleja la realidad del negocio: a partir de cierta cantidad de llamadas, el cliente ya está saturado. Para el modelo, el efecto de llamar 14 veces o 56 veces es el mismo (un rotundo "no").

**En resumen:** Se usa la Winsorización por Percentiles porque es el **único método** que estabiliza el dataset para el modelo espacial (SVM), sobrevive a la extrema concentración de ceros en el historial de llamadas, y retiene a los clientes atípicos que son cruciales para entender el comportamiento de la campaña de marketing.

En el contexto específico de tu dataset bancario y las variables que estamos tratando, la Winsorización suele ser preferida por los equipos de ciencia de datos debido a un factor crucial: **el sentido de negocio y la interpretabilidad.**

Aunque la transformación logarítmica (`log1p`) es matemáticamente más "suave", la Winsorización resuelve problemas prácticos que el logaritmo distorsiona. Aquí te presento la justificación de por qué es totalmente válido (y a menudo mejor) quedarte con la Winsorización para este proyecto:

### 1. El significado lógico de las variables (El caso de la Edad)
Aplicar un logaritmo a variables financieras (como saldo en cuenta o ingresos) es una regla de oro, pero aplicarlo a la variable `age` (edad) no tiene mucho sentido.
* Si usas `log1p` en la edad, comprimes la diferencia entre un joven de 20 años y un adulto de 40, distorsionando el ciclo de vida real del cliente.
* Con la **Winsorización**, el cliente de 98 años simplemente se "topa" a 71 años (el percentil 99). Para el modelo SVM, esto se traduce lógicamente como: *"trata a este cliente como a cualquier otro del grupo de la tercera edad"*, manteniendo intacta la escala original (en años) para el 99% restante de tu base.

### 2. El límite de tolerancia del cliente (El caso de la Campaña)
La variable `campaign` mide cuántas veces llamaste a una persona.
* El logaritmo reducirá un valor atípico de 56 llamadas a un número como ~4.0. Matemáticamente funciona, pero se pierde el concepto real.
* La **Winsorización** topa ese 56 en, digamos, 14 llamadas. Esto empata perfectamente con la lógica de marketing: a partir de la llamada número 14, el cliente ya está saturado. El efecto de llamarlo 14 veces o 56 veces es exactamente el mismo (un rechazo rotundo). Toparlo refleja esa "barrera de saturación" real.

### 3. La redundancia en variables dominadas por ceros
Como vimos, la variable `previous` (contactos antes de esta campaña) está en 0 para el 85% de los clientes. El valor máximo es 7.
* Si le aplicas `log1p` a un 7, se convierte en ~2.07.
* Realmente, la distancia espacial entre 0 y 7 no es tan masiva como para deformar críticamente tu modelo SVM. Topar ese 7 a un 3 o 4 (Winsorización) es más que suficiente para estabilizar el hiperplano sin tener que transformar toda la columna a escalas logarítmicas confusas.

### Resumen
El logaritmo es excelente cuando la variable crece de forma exponencial (como el uso de datos en gigabytes de una telefónica o el balance en dólares de una cuenta). Pero para **conteos pequeños y edades**, la Winsorización por percentiles es la herramienta más precisa porque protege la matemática de tu SVM sin destruir la interpretabilidad de lo que cada número significa para el banco.

Sabiendo que ambos métodos son válidos pero tienen impactos distintos en la interpretabilidad, ¿prefieres que definamos la Winsorización como la técnica oficial para tu pipeline de preprocesamiento, o te gustaría que configuremos un experimento para probar ambas y ver cuál le otorga mejor rendimiento (F1-Score) al SVM?

La gráfica revela un problema muy claro (y clásico) con las variables macroeconómicas de este dataset.

Aquí tienes las conclusiones principales y cómo impactan tu proyecto:

### 1. El bloque de alta multicolinealidad
El mapa de calor muestra un bloque de correlación positiva extremadamente fuerte (tonos rojos intensos, con valores superiores a 0.90) entre tres variables principales:
* **`emp.var.rate`** (Tasa de variación del empleo)
* **`euribor3m`** (Tasa de interés a 3 meses)
* **`nr.employed`** (Número de empleados)

**¿Por qué sucede esto?**
Estas tres variables describen exactamente el mismo fenómeno: la salud de la economía europea en el momento de la llamada. Cuando la economía está en expansión, hay más empleados, la tasa de variación es positiva y los bancos centrales suben el Euribor para controlar la inflación. Si dejas las tres, le estás dando al modelo la misma información repetida tres veces.

*(Nota: `cons.price.idx` también muestra una correlación moderada-alta con este grupo, alrededor de 0.70 - 0.80, lo que refuerza esta redundancia).*

### 2. La relación `pdays` y `previous`
También se observa una correlación negativa moderada (el cuadro azul oscuro, alrededor de -0.60) entre `pdays` (días desde el último contacto) y `previous` (número de contactos previos).
Tiene lógica: si a un cliente se le ha contactado muchas veces (`previous` alto), es probable que el tiempo transcurrido desde el último contacto (`pdays`) sea menor. Sin embargo, esta correlación no es lo suficientemente severa (no supera el 0.85) como para considerarla un problema crítico de multicolinealidad.

### 3. Impacto en tus modelos
Como platicamos anteriormente, la multicolinealidad afecta a los modelos de forma diferente:
* **Para tu Máquina de Soporte Vectorial (SVM) y Regresión Logística:** Esto es **tóxico**. La multicolinealidad desestabiliza las matemáticas detrás de la creación del hiperplano, haciendo que los pesos de las variables se vuelvan inestables y el modelo sea propenso al sobreajuste (overfitting).
* **Para los Bosques Aleatorios (Random Forest):** No afectará su capacidad predictiva, pero arruinará el gráfico de "Importancia de Variables" (*Feature Importance*), ya que el árbol dividirá el crédito de la importancia entre las tres variables económicas de forma aleatoria, dificultando la interpretación de negocio.

# Selección de variables

**Variables categóricas** (job, marital, etc.): Debes llamarla con discretize=False.

**Variables numéricas** (age, campaign): Debes usar discretize=True.

```python
def woe_iv(df, feature, target, bins=5, discretize=False):
    eps = 0.00001
    df = df.copy()

    # 1. Aseguramos que el target sea binario numérico o esté bien ordenado
    # Forzamos que 'no' sea 0 y 'yes' sea 1 para consistencia
    if df[target].dtype == 'object':
        df[target] = df[target].map({'no': 0, 'yes': 1})

    # 2. Discretizar variables numéricas
    if discretize:
        # Usamos qcut para crear bins con igual cantidad de registros
        df[feature] = pd.qcut(df[feature], q=bins, duplicates="drop").astype(str)

    # 3. Calculamos la tabla cruzada
    # Al ser 0 y 1, el crosstab pondrá 0 en la primera col y 1 en la segunda
    tmp = pd.crosstab(df[feature], df[target], normalize='columns')

    tmp.columns = ["%no_contrata", "%si_contrata"]

    # 4. Cálculo de WoE e IV
    tmp["WoE"] = np.log((tmp["%si_contrata"] + eps) / (tmp["%no_contrata"] + eps))
    tmp["IV_contrib"] = (tmp["%si_contrata"] - tmp["%no_contrata"]) * tmp["WoE"]

    iv = tmp["IV_contrib"].sum()
    return tmp, iv
```

### woe_numericas

```python


def calcular_bloque_iv(df, lista_variables, target, tipo_var):
    """
    Automatiza el cálculo de IV para un grupo de variables.
    tipo_var: 'num' para numéricas (aplica discretización)
              'cat' para categóricas
    """
    resultados = []

    # Determinamos si se debe discretizar basado en el parámetro tipo_var
    debe_discretizar = True if tipo_var == 'num' else False

    for var in lista_variables:
        try:
            # Llamamos a la función base woe_iv que ya definiste
            _, iv_valor = woe_iv(df, var, target, bins=5, discretize=debe_discretizar)

            # Clasificamos el poder predictivo según la escala estándar
            if iv_valor < 0.02: poder = "Inútil"
            elif iv_valor < 0.1: poder = "Débil"
            elif iv_valor < 0.3: poder = "Medio"
            elif iv_valor < 0.5: poder = "Fuerte"
            else: poder = "Muy Fuerte / Sospechoso"

            resultados.append({
                'Variable': var,
                'IV': iv_valor,
                'Poder_Predictivo': poder,
                'Tipo': tipo_var
            })
        except Exception as e:
            print(f"Error procesando {var}: {e}")

    return pd.DataFrame(resultados).sort_values(by='IV', ascending=False)
```

Aunque aplicamos Winsorización a variables como age o campaign, siempre pueden quedar valores en los extremos.

MinMaxScaler: Es extremadamente sensible a los valores extremos. Si tienes un solo cliente de 100 años, el resto de los clientes de 20 a 40 años quedarían "comprimidos" en un rango minúsculo (ej. entre 0.1 y 0.2), perdiendo mucha información.

StandardScaler: Al usar la media y la desviación estándar, es mucho más robusto. Los valores extremos no "colapsan" al resto de la distribución; simplemente quedan representados como valores a 3 o 4 desviaciones estándar del centro.

Como tu dataset tiene variables con distribuciones muy distintas (la edad, el índice macroeconómico y las variables binarias del encoding), el StandardScaler nivela el campo de juego sin perder la riqueza de la varianza que el SVM necesita para decidir quién contratará un depósito y quién no.

# SVM

```python
from sklearn.svm import SVC
from sklearn.model_selection import cross_validate, StratifiedKFold
from sklearn.pipeline import make_pipeline

# 1. Definimos la estrategia de validación
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# 2. Creamos un pipeline que escala y luego aplica el SVM
# Usamos class_weight='balanced' para compensar el desbalanceo
svm_pipe = make_pipeline(
    StandardScaler(),
    SVC(kernel='rbf', C=1.0, probability=True, class_weight='balanced', random_state=42)
)

# 3. Ejecutamos la validación cruzada
metrics = ['accuracy', 'recall', 'roc_auc']
cv_results_svm = cross_validate(svm_pipe, X_train, y_train, cv=skf, scoring=metrics, n_jobs=-1)

print(f"AUC Promedio SVM: {cv_results_svm['test_roc_auc'].mean():.4f}")
print(f"Sensibilidad Promedio SVM: {cv_results_svm['test_recall'].mean():.4f}")
```

```python
# Entrenamos el modelo final
svm_pipe.fit(X_train, y_train)

# Obtenemos probabilidades y calculamos curva ROC
y_probs_svm = svm_pipe.predict_proba(X_test)[:, 1]
fpr_svm, tpr_svm, thresholds_svm = roc_curve(y_test, y_probs_svm)
auc_svm = roc_auc_score(y_test, y_probs_svm)

# Índice de Youden para SVM
youden_svm = tpr_svm - fpr_svm
ix_svm = np.argmax(youden_svm)
best_thresh_svm = thresholds_svm[ix_svm]

print(f"Mejor Umbral SVM (Youden): {best_thresh_svm:.4f}")
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

# 1. Obtener las probabilidades de la clase positiva (1)
y_probs_svm = svm_pipe.predict_proba(X_test)[:, 1]

# 2. Aplicar el umbral de Youden (0.1579)
# Si la probabilidad es mayor o igual al umbral, se clasifica como 1 (Sí)
y_pred_svm_custom = (y_probs_svm >= 0.1579).astype(int)

# 3. Crear la matriz de confusión
cm_svm = confusion_matrix(y_test, y_pred_svm_custom)

# 4. Graficar
fig, ax = plt.subplots(figsize=(8, 6))
disp_svm = ConfusionMatrixDisplay(confusion_matrix=cm_svm, display_labels=['No Contrata', 'Sí Contrata'])

disp_svm.plot(cmap='Purples', ax=ax, values_format='d') # Usamos tonos púrpuras para diferenciarlo del RF
plt.title('Matriz de Confusión SVM', fontsize=14)
plt.show()

# 5. Análisis de eficiencia
tn_s, fp_s, fn_s, tp_s = cm_svm.ravel()
recall_custom = tp_s / (tp_s + fn_s)
espec_custom = tn_s / (tn_s + fp_s)

print(f"--- ANÁLISIS ESTRATÉGICO SVM (UMBRAL CUSTOM) ---")
print(f"Sensibilidad (Recall): {recall_custom:.4f}")
print(f"Especificidad: {espec_custom:.4f}")
print(f"Falsos Negativos (Clientes perdidos): {fn_s}")
```

```python
from sklearn.metrics import roc_curve, roc_auc_score
import matplotlib.pyplot as plt
import numpy as np

# 1. Obtener probabilidades para el set de prueba
y_probs_svm = svm_pipe.predict_proba(X_test)[:, 1]

# 2. Calcular métricas de la curva
fpr_svm, tpr_svm, thresholds_svm = roc_curve(y_test, y_probs_svm)
auc_svm = roc_auc_score(y_test, y_probs_svm)

# 3. Identificar el punto de Youden en la curva
youden_svm = tpr_svm - fpr_svm
ix_svm = np.argmax(youden_svm)
best_thresh_svm = thresholds_svm[ix_svm]

# 4. Graficar
plt.figure(figsize=(9, 7))
plt.plot(fpr_svm, tpr_svm, color='mediumpurple', lw=3,
         label=f'SVM RBF (AUC = {auc_svm:.3f})')
plt.plot([0, 1], [0, 1], color='navy', lw=1, linestyle='--', label='Azar')

# Resaltar el punto de Youden del SVM
plt.scatter(fpr_svm[ix_svm], tpr_svm[ix_svm], marker='o', color='black',
            label=f'Punto Óptimo (Youden)', zorder=5)

# Anotación técnica
plt.annotate(f'Umbral Óptimo: {best_thresh_svm:.4f}\nSens: {tpr_svm[ix_svm]:.2f}, Esp: {1-fpr_svm[ix_svm]:.2f}',
             xy=(fpr_svm[ix_svm], tpr_svm[ix_svm]), xytext=(fpr_svm[ix_svm]+0.1, tpr_svm[ix_svm]-0.15),
             arrowprops=dict(facecolor='black', shrink=0.05, width=1, headwidth=8))

plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('Tasa de Falsos Positivos (1 - Especificidad)')
plt.ylabel('Tasa de Verdaderos Positivos (Sensibilidad)')
plt.title('Curva ROC - Support Vector Machine (SVM)')
plt.legend(loc="lower right")
plt.grid(alpha=0.3)
plt.show()

print(f"Área bajo la curva SVM (AUC): {auc_svm:.4f}")
```

### 2. Comportamiento Algorítmico por Modelo

#### A. Random Forest (Ensamblado por Bagging)
* **Comportamiento:** Como modelo base, el Random Forest promedió las predicciones de múltiples árboles de decisión independientes. Al evaluarse con el umbral por defecto (0.50), el modelo sufrió las consecuencias directas del desbalanceo de clases, sesgándose fuertemente hacia la clase mayoritaria.
* **Diagnóstico Estadístico:** Alcanzó la Exactitud (Accuracy) más alta (0.8861) y la mejor Especificidad (0.9592), pero a costa de una Sensibilidad deficiente (0.3023). El algoritmo fue incapaz de identificar patrones subyacentes complejos de la clase minoritaria, clasificando casi todo como negativo. Su AUC de 0.7619 establece un rendimiento predictivo moderado.

#### B. Support Vector Machine (Kernel RBF)
* **Comportamiento:** El SVM transformó el espacio de características a una dimensión superior utilizando un kernel de base radial (RBF) para encontrar un hiperplano de separación no lineal. Se compensó el desbalanceo mediante pesos de clase y se calibró la frontera de decisión utilizando el punto óptimo de Youden.
* **Diagnóstico Estadístico:** El ajuste del hiperplano mediante el umbral de 0.1579 forzó al modelo a aumentar la Tasa de Verdaderos Positivos (Sensibilidad del 0.6210). Esto demuestra que el mapeo no lineal logró encontrar instancias positivas que los cortes ortogonales de los árboles de decisión no detectaron. Sin embargo, la varianza aumentó, lo que se tradujo en una caída de la Especificidad a 0.8588 y un aumento notable en los Falsos Positivos. El AUC mejoró a 0.7762.

#### C. XGBoost (Ensamblado por Gradient Boosting)
* **Comportamiento:** La arquitectura secuencial de XGBoost permitió minimizar la función de pérdida iterativamente (logloss), donde cada nuevo árbol corrigió los residuales del anterior. El parámetro `scale_pos_weight` y la optimización del umbral (0.5237) permitieron una adaptación dinámica a la distribución asimétrica de los datos.
* **Diagnóstico Estadístico:** XGBoost exhibió la mayor capacidad de generalización y separabilidad intrínseca, siendo el único algoritmo en superar el umbral de 0.80 en AUC (0.8016). Técnicamente, logró dominar el *trade-off* entre sesgo y varianza mejor que el SVM: maximizó la Tasa de Verdaderos Positivos (0.6288) al mismo tiempo que redujo la Tasa de Falsos Positivos en comparación con el modelo geométrico.

Desde la perspectiva del rendimiento algorítmico, **XGBoost es el modelo matemáticamente superior**.

La métrica de Exactitud (Accuracy) resultó ser una medida engañosa debido al sesgo hacia la clase mayoritaria (Paradoja de la Exactitud), por lo que la decisión se fundamenta en el Área bajo la Curva ROC. El incremento del AUC a 0.8016 por parte de XGBoost, respaldado por una fuerte consistencia observada en la Validación Cruzada (0.7920), indica que el descenso de gradiente secuencial logró modelar de forma más precisa la frontera de decisión entre ambas clases que la hiper-parametrización espacial del SVM o el paralelismo del Random Forest. Su equilibrio superior entre TPR (Sensibilidad) y FPR (1 - Especificidad) lo consolida como el clasificador más robusto y calibrado del estudio.
